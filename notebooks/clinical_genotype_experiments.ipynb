## 2. Target Variable Analysis

```python
# Identify target column (adjust based on actual column name)
# Common names: 'diagnosis', 'DX', 'label', 'target', 'Group'
target_candidates = ['diagnosis', 'DX', 'label', 'target', 'Group', 'DIAGNOSIS']

for col in target_candidates:
    if col in balanced_df.columns:
        print(f"Found target column: {col}")
        print(f"\nValue counts:\n{balanced_df[col].value_counts()}")
        break
```

```python
# Visualize target distribution
# Adjust 'target_col' based on actual column name found above
target_col = 'diagnosis'  # UPDATE THIS based on your data

if target_col in balanced_df.columns:
    fig, ax = plt.subplots(figsize=(10, 6))
    balanced_df[target_col].value_counts().plot(kind='bar', ax=ax, color='steelblue', edgecolor='black')
    ax.set_title('Distribution of Diagnosis Classes', fontsize=14)
    ax.set_xlabel('Diagnosis', fontsize=12)
    ax.set_ylabel('Count', fontsize=12)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```

## 3. Clinical Features Analysis

```python
# Identify clinical features (CDR domains, cognitive scores, etc.)
# Common clinical features: CDRSB, MMSE, ADAS, FAQ, etc.
clinical_keywords = ['CDR', 'MMSE', 'ADAS', 'FAQ', 'Memory', 'Orientation', 'Judgment']

clinical_cols = [col for col in balanced_df.columns 
                 if any(kw.lower() in col.lower() for kw in clinical_keywords)]

print(f"Potential clinical columns found: {len(clinical_cols)}")
print(clinical_cols)
```

```python
# Correlation analysis of clinical features
if len(clinical_cols) > 1:
    clinical_data = balanced_df[clinical_cols].select_dtypes(include=[np.number])
    
    if not clinical_data.empty:
        fig, ax = plt.subplots(figsize=(12, 10))
        corr_matrix = clinical_data.corr()
        sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, 
                    fmt='.2f', ax=ax, square=True)
        ax.set_title('Correlation Matrix of Clinical Features', fontsize=14)
        plt.tight_layout()
        plt.show()
```

## 4. Genotype (APOE) Analysis

```python
# Look for APOE-related columns
apoe_keywords = ['APOE', 'genotype', 'allele', 'e2', 'e3', 'e4']

apoe_cols_balanced = [col for col in balanced_df.columns 
                      if any(kw.lower() in col.lower() for kw in apoe_keywords)]
apoe_cols_apoe = [col for col in apoe_df.columns 
                  if any(kw.lower() in col.lower() for kw in apoe_keywords)]

print(f"APOE columns in balanced dataset: {apoe_cols_balanced}")
print(f"APOE columns in APOE dataset: {apoe_cols_apoe}")
```

```python
# APOE genotype distribution
# Adjust column name based on actual data
apoe_col = None  # UPDATE THIS based on your data

for col in apoe_cols_balanced + apoe_cols_apoe:
    if 'genotype' in col.lower() or 'apoe' in col.lower():
        apoe_col = col
        break

if apoe_col and apoe_col in balanced_df.columns:
    print(f"APOE genotype distribution:")
    print(balanced_df[apoe_col].value_counts())
```
